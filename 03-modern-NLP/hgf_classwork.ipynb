```python
from transformers import pipeline
```

### Huggingface - Basic Inference

```python
model = pipeline("sentiment-analysis")
```

```python
model("Today is an exciting day for learning AI")
```

```python
model("Today is a great day for learning AI but I will go out")
```

### Zero Shot Classification

```python
model = pipeline('zero-shot-classification')
```

```python
model("This is a course on Huggingface", candidate_labels=["sports", "politics", "education", "businees"])
```

```python
model("Reliance Industries stock rose sharply today", candidate_labels=["sports", "politics", "education", "business"])
```

### Use a specific model

```python
model = pipeline('text-generation', model="HuggingFaceTB/SmolLM2-360M")
```

```python
text = model("Complete the statement 'Gravity is a force ...'") 
```

```python
print(text[0]['generated_text'])
```

### Wrong Use-Case

```python
model('classify the text into ["politics", "sports"]: "India won the world cup match')
```
